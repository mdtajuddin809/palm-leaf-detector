```python
import os

input_folder = "All Images Detection"
output_folder = "runs/detect/predict"

processed = set(os.listdir(output_folder))  # jo images already ho chuki hain
```

```python
remaining_images = []

for img in os.listdir(input_folder):
    if img not in processed:
        remaining_images.append(img)

print("Remaining images:", len(remaining_images))
```

```python
from ultralytics import YOLO
import os

model = YOLO("best.pt")

for img in remaining_images:
    path = os.path.join(input_folder, img)

    model.predict(
        source=path,
        imgsz=640,
        batch=1,
        device="cpu",
        save=True,
        save_txt=True
    )
```

```python
import os
import shutil

folders = [
    "runs/detect/predict",
    "runs/detect/predict2"
]

final_images = "final_output/images"
final_labels = "final_output/labels"

os.makedirs(final_images, exist_ok=True)
os.makedirs(final_labels, exist_ok=True)

for folder in folders:
    # Images copy
    for file in os.listdir(folder):
        if file.endswith(".jpg"):
            shutil.copy(os.path.join(folder, file), final_images)
    
    # Labels copy
    label_path = os.path.join(folder, "labels")
    if os.path.exists(label_path):
        for file in os.listdir(label_path):
            shutil.copy(os.path.join(label_path, file), final_labels)

print("✅ Merge Done!")
```

```python
import os

images_folder = "final_output/images"
labels_folder = "final_output/labels"

images = set(os.path.splitext(f)[0] for f in os.listdir(images_folder) if f.endswith(".jpg"))
labels = set(os.path.splitext(f)[0] for f in os.listdir(labels_folder) if f.endswith(".txt"))

missing = images - labels
print("Missing labels:", missing)

for name in missing:
    open(os.path.join(labels_folder, name + ".txt"), "w").close()

print("Empty label files created.")
```
